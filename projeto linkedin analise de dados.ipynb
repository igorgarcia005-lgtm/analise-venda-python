```python
df_vendas['cidade']=df_vendas['cidade'].fillna('Não Informado') # Preenchendo valores nulos na coluna cidade com "Não Informado"

mediana_idade = df_vendas['Idade Cliente'].median() # Calculando a mediana da coluna idade_cliente
df_vendas['Idade Cliente'] = df_vendas['Idade Cliente'].fillna(mediana_idade)



# Verifica se ainda sobrou algum NaN nas colunas cidade e Idade Cliente
print(df_vendas[['cidade', 'Idade Cliente']].isnull().sum())# Preenchendo valores nulos na coluna idade_cliente com a mediana calculada
```

```python
colunas_financeiras = ['Valor Unitario', 'Valor Total']

for coluna in colunas_financeiras:
  # Converte diretamente para numérico. Se houver algum texto inválido, vira NaN de forma segura.
  df_vendas[coluna] = pd.to_numeric(df_vendas[coluna], errors='coerce')

# Confere os tipos e se sobrou algum valor nulo
print(df_vendas[['Valor Unitario', 'Valor Total']].dtypes)
print(df_vendas[['Valor Unitario', 'Valor Total']].isnull().sum())
```

```python
# 1. Total de valores ausentes em todas as colunas da base
print(df_vendas.isnull().sum())

# 2. Total de registros duplicados na base inteira
print(f"Total de linhas duplicadas: {df_vendas.duplicated().sum()}")
```

```python
condicao_vt = (
    df_vendas['Valor Total'].isnull() #mostrando a condição para calcular o valor total quando ele estiver faltando, mas temos o valor unitário e a quantidade
    & df_vendas['Valor Unitario'].notnull() #mostrando a condição para calcular o valor total quando ele estiver faltando, mas temos o valor unitário e a quantidade
    & df_vendas['quantidade'].notnull() #mostrando a condição para calcular o valor total quando ele estiver faltando, mas temos o valor unitário e a quantidade
)
df_vendas.loc[condicao_vt, 'Valor Total'] = (
    df_vendas.loc[condicao_vt, 'Valor Unitario'] #utilizando a função loc para localizar as linhas que atendem à condição e calcular o valor total multiplicando o valor unitário pela quantidade   
    * df_vendas.loc[condicao_vt, 'quantidade']
)

# 2. Se o 'Valor Unitario' está faltando, mas temos 'Valor Total' e 'Quantidade'
condicao_vu = (
    df_vendas['Valor Unitario'].isnull() #mostrando a condição para calcular o valor unitário quando ele estiver faltando, mas temos o valor total e a quantidade
    & df_vendas['Valor Total'].notnull() #mostrando a condição para calcular o valor unitário quando ele estiver faltando, mas temos o valor total e a quantidade
    & df_vendas['quantidade'].notnull() #mostrando a condição para calcular o valor unitário quando ele estiver faltando, mas temos o valor total e a quantidade
    & (df_vendas['quantidade'] > 0)
)
df_vendas.loc[condicao_vu, 'Valor Unitario'] = (
    df_vendas.loc[condicao_vu, 'Valor Total']
    / df_vendas.loc[condicao_vu, 'quantidade']  # ou .loc[condicao_vu, 'Quantidade']
)

# 3. Verifique quantos nulos restaram após o recálculo
print(df_vendas[['Valor Unitario', 'Valor Total']].isnull().sum())

df_vendas = df_vendas.dropna(subset=['Valor Unitario', 'Valor Total'])
print(df_vendas[['Valor Unitario', 'Valor Total']].isnull().sum())
```
